## Problems 1–2: Functions

```python
A={1,2,3}; B={'a','b','c'}; f={1:'a',2:'b',3:'c'}
print('Bijective example:', f, function_properties(A,B,f))
injective_nononto=[]
for values in product(sorted(B),repeat=3):
    candidate=dict(zip(sorted(A),values))
    if function_properties(A,B,candidate)==(True,False): injective_nononto.append(candidate)
print('Injective but non-surjective A→B functions:', injective_nononto)
print('Larger-codomain example:', function_properties(A,{'a','b','c','d'},f))
```

## Problems 3–7: Relation Properties

```python
R={(1,3),(1,2),(2,2)}; S={(1,2),(2,3),(3,1)}
R4={(1,1),(1,3),(2,1),(2,3),(3,3)}
for name,relation in [('R',R),('S',S),('Problem 4 R',R4)]:
    print(name, relation_properties({1,2,3},relation), '
')
people={'Alex','Blair','Casey'}
siblings={('Alex','Blair'),('Blair','Alex'),('Blair','Casey'),('Casey','Blair')}
print('Sibling model:', relation_properties(people,siblings))
```

## Problems 8–9: Equivalence Modulo 6

```python
sample=set(range(-12,13))
mod6={(x,y) for x in sample for y in sample if (x-y)%6==0}
print(relation_properties(sample,mod6))
for remainder in range(6):
    print(f'[{remainder}] =', sorted(x for x in sample if x%6==remainder))
```

## Problems 14–16: Recursive Sum Complexity

```python
for n in (0,1,5,10,25,100):
    data=list(range(1,n+1)); count={'calls':0}
    answer=recursive_sum(data,counter=count)
    print(f'n={n:3}, sum={answer:5}, calls={count["calls"]:3}, expected depth={n+1:3}')
print('Time: T(n)=T(n-1)+Θ(1)=Θ(n)')
print('Recursive space: Θ(n); iterative auxiliary space: Θ(1)')
```
